## Synthetic Example

```python
# Create synthetic power spectrum
ell = np.logspace(2, 4, 50)
D_ell_theory = 1e-12 * (ell/1000)**(-0.5)

# Add some scatter to simulate empirical measurements
np.random.seed(42)
D_ell_empirical = D_ell_theory * (1 + 0.1 * np.random.randn(len(ell)))
D_ell_error = 0.05 * D_ell_theory

plt.figure(figsize=(10, 6))
plt.errorbar(ell, D_ell_empirical, yerr=D_ell_error, 
             fmt='o', capsize=3, alpha=0.7, label='Empirical')
plt.loglog(ell, D_ell_theory, '--', label='Theory', linewidth=2)
plt.xlabel(r'$\ell$')
plt.ylabel(r'$D_\ell$ [μK²]')
plt.title('Power Spectrum: Theory vs Empirical')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
